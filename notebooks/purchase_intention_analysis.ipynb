# Factors Influencing Online Purchase Intention Among Pakistani Consumers

**Synthetic demo mode.** Replace the demo CSV with the original anonymized survey data before reporting empirical results.

```python
from pathlib import Path
import pandas as pd
import statsmodels.formula.api as smf
df=pd.read_csv(Path('..')/'data/purchase_intention_data_demo.csv')
df.head()
```

## 1. Data validation

```python
numeric=['perceived_risk','payment_trust','delivery_reliability','product_information','convenience','digital_interaction','perceived_value','purchase_intention']
for c in numeric: df[c]=pd.to_numeric(df[c],errors='coerce')
print('Rows:',len(df))
display(df.isna().sum().sort_values(ascending=False).to_frame('missing'))
```

## 2. Descriptive statistics

```python
complete=df.dropna(subset=numeric).copy()
display(complete[numeric].describe().T.round(2))
```

## 3. Spearman correlations

```python
corr=complete[numeric].corr(method='spearman')
display(corr.round(3))
```

## 4. OLS regression with HC3 robust standard errors

```python
model=smf.ols('purchase_intention ~ perceived_risk + payment_trust + delivery_reliability',data=complete).fit(cov_type='HC3')
print(model.summary())
```
